```python
# ============================================================
# BRIDGE LOAD DISTRIBUTION CALCULATOR
# B.Tech Mathematics Mini Project - A2
# ============================================================

"""
Mathematical Model
------------------
We consider a continuous Euler-Bernoulli beam with 3 vertical supports:

Support 1: x = 0
Support 2: x = S/2
Support 3: x = S

Unknown reactions:
R1, R2, R3

We use:
1. Vertical force equilibrium
2. Moment equilibrium
3. Compatibility at the middle support

The three equations are written as:

        A X = B

and solved using:

        X = A^(-1) B

where X = [R1, R2, R3]^T.

This is an educational model and is not intended for actual bridge design.
"""

print("Bridge Load Distribution Calculator")
print("Mathematical model: AX = B")
print("Three supports -> three reaction forces")
```

```python
# ============================================================
# IMPORT LIBRARIES
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from IPython.display import display
```

```python
# ============================================================
# INPUT CONFIGURATION
# ============================================================

# Bridge total span in metres
SPAN = 20.0

# Beam flexural rigidity EI
# For this educational model, use a convenient value.
# Units are consistent with kN and metres.
EI = 1.0e7

# ------------------------------------------------------------
# FIXED LOADS
# Format:
# {"name": ..., "position": ..., "load": ...}
# Load is in kN and position is in metres.
# ------------------------------------------------------------

fixed_loads = [
    {
        "name": "Fixed Load 1",
        "position": 4.0,
        "load": 100.0
    },
    {
        "name": "Fixed Load 2",
        "position": 15.0,
        "load": 80.0
    }
]

# ------------------------------------------------------------
# MOVING TRUCK
# ------------------------------------------------------------

truck_load = 150.0

truck_position_before = 6.0
truck_position_after = 12.0

print("Bridge span:", SPAN, "m")
print("Truck load:", truck_load, "kN")
print("Truck position before:", truck_position_before, "m")
print("Truck position after :", truck_position_after, "m")
```

```python
# ============================================================
# SUPPORT POSITIONS
# ============================================================

support_positions = np.array([
    0.0,
    SPAN / 2,
    SPAN
])

print("Support positions:")
print("Support 1 =", support_positions[0], "m")
print("Support 2 =", support_positions[1], "m")
print("Support 3 =", support_positions[2], "m")
```

```python
# ============================================================
# DEFLECTION / FLEXIBILITY FUNCTION
# ============================================================

def midspan_flexibility(load_position, span, EI):
    """
    Calculate the downward deflection at the middle of a
    simply-supported beam caused by a unit downward point load.

    Parameters
    ----------
    load_position : float
        Position of the point load (m)

    span : float
        Total beam span (m)

    EI : float
        Flexural rigidity

    Returns
    -------
    float
        Flexibility coefficient
    """

    a = span / 2
    b = load_position
    L = span

    # Load is on the left half
    if b <= a:

        flexibility = (
            a * (L - b) *
            (2 * b * L - b**2 - a**2)
            / (6 * L * EI)
        )

    # Load is on the right half
    else:

        flexibility = (
            b * (L - a) *
            (2 * a * L - a**2 - b**2)
            / (6 * L * EI)
        )

    return flexibility
```

```python
# ============================================================
# BUILD AX = B
# ============================================================

def build_matrix_system(loads, span, EI):
    """
    Construct the 3x3 matrix system:

        A X = B

    Unknown vector:

        X = [R1, R2, R3]^T

    Equations:

    1. Vertical force equilibrium
       R1 + R2 + R3 = total load

    2. Moment equilibrium about Support 1
       (S/2)R2 + S R3 = sum(P_i x_i)

    3. Compatibility at middle support
       f_mid R2 = sum(P_i f_i)

    """

    middle = span / 2

    # --------------------------------------------------------
    # Matrix A
    # --------------------------------------------------------

    flexibility_middle = midspan_flexibility(
        middle,
        span,
        EI
    )

    A = np.array([
        [1.0, 1.0, 1.0],
        [0.0, middle, span],
        [0.0, flexibility_middle, 0.0]
    ])

    # --------------------------------------------------------
    # Vector B
    # --------------------------------------------------------

    total_load = sum(loads.values())

    total_moment = sum(
        load * position
        for position, load in loads.items()
    )

    load_deflection = sum(
        load * midspan_flexibility(
            position,
            span,
            EI
        )
        for position, load in loads.items()
    )

    B = np.array([
        total_load,
        total_moment,
        load_deflection
    ])

    return A, B
```

```python
# ============================================================
# SOLVE AX = B USING MATRIX INVERSE
# ============================================================

def solve_bridge(loads, span, EI):

    A, B = build_matrix_system(loads,span,EI)


    # Check whether a unique solution exists
    determinant = np.linalg.det(A)

    if abs(determinant) < 1e-12:
        raise ValueError(
            "Matrix A is singular. A unique solution does not exist."
        )

    # Matrix inverse
    A_inverse = np.linalg.inv(A)

    # X = A^(-1) B
    X = A_inverse @ B

    return A, B, determinant, A_inverse, X
```

```python

# ============================================================
# SOLVE BEFORE AND AFTER TRUCK MOVEMENT
# ============================================================

# Create loads dictionaries (key: position, value: load magnitude)
loads_before = {
    load["position"]: load["load"] for load in fixed_loads
}
loads_before[truck_position_before] = truck_load

loads_after = {
    load["position"]: load["load"] for load in fixed_loads
}
loads_after[truck_position_after] = truck_load

A_before, B_before, det_before, Ainv_before, R_before = solve_bridge(
    loads_before,
    SPAN,
    EI
)

A_after, B_after, det_after, Ainv_after, R_after = solve_bridge(
    loads_after,
    SPAN,
    EI
)

print("Calculation completed successfully.")
```
